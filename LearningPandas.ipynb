Creating series


```python
import numpy as np
import pandas as pd
```

```python
series = pd.Series([1,2,3,4,5])
print(series)
```

```python
series_with_index = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(series_with_index)
```

```python
series_with_dictionary = pd.Series({'1':1,'2':2,'3':3})
print(series_with_dictionary)
```

Creating a dataframe from dictionary

```python
data = {
  'text':['I love AI', 'Machine learing is great','Gen AI is the future'],
  'label':[1,1,1],
  'length':[9,24,20]
}
df = pd.DataFrame(data)
print(df)
```

creating a dataframe from list

```python
data_list = [['I love AI',1,9],['Machine learing is great',1,24],['Gen AI is the future',1,20]]
df = pd.DataFrame(data_list,columns=['text','label','length'])
print(df)
```

Basic DataFrame Attributes

```python
df = pd.DataFrame({
    'text':['A','B','C','D','E'],
    'label':[1,1,1,1,1],
    'length':[1,2,3,4,5]
})

print(df.shape)
print(df.columns)
print(df.index)
print(df.dtypes)
```
