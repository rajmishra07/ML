```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
Sales_data = np.array([
    [1,150000,180000,220000,250000],
    [2,120000,140000,160000,190000],
    [3,200000,230000,260000,300000],
    [4,180000,210000,240000,270000],
    [5,160000,185000,205000,230000]
])
print("==============Company sales analysis======================")
print("sales data shape :", Sales_data.shape)
print("sample data for First 3 restaurent :", Sales_data[0:3])
```

Total Sales per year

```python
print(np.sum(Sales_data, axis=0))
total_sales_per_year = np.sum(Sales_data[:,1:],axis=0)
print(total_sales_per_year)
```

min sale per restaurant

```python
min_sale = np.min(Sales_data[:,1:],axis=1)
print(min_sale)
```

max sale per year

```python
max_sale = np.max(Sales_data[:,1:],axis=0)
print(max_sale)
```

Avg Sale per restaurent

```python
avg_sale = np.mean(Sales_data[:,1:],axis=1)
print(avg_sale)
```

cumulative  Sale

```python
cumsum = np.cumsum(Sales_data[:,1:],axis=1)
print(cumsum)
```
